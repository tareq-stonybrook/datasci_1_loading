## Breast Cancer Wisconsin Dataset

Dataset contains cell nuclei measurements for 699 breast tissue samples used to classify tumors as benign or malignant.

## Columns

1. **Id**: sample identification number
2. **Cl.thickness**: clump thickness (1 to 10)
3. **Cell.size**: uniformity of cell size (1 to 10)
4. **Cell.shape**: uniformity of cell shape (1 to 10)
5. **Marg.adhesion**: marginal adhesion (1 to 10)
6. **Epith.c.size**: single epithelial cell size (1 to 10)
7. **Bare.nuclei**: bare nuclei count (1 to 10)
8. **Bl.cromatin**: bland chromatin (1 to 10)
9. **Normal.nucleoli**: normal nucleoli (1 to 10)
10. **Mitoses**: mitoses (1 to 10)
11. **Class**: 0 indicates benign, 1 indicates malignant
