## Real Estate Valuation

Market historical data of real estate valuation collected from Sindian District, New Taipei City, Taiwan. Used for predicting house price per unit area based on location and property features.

## Columns

* No : row index
* X1 transaction date : transaction date as decimal year, e.g. 2012.917
* X2 house age : house age in years
* X3 distance to the nearest MRT station : distance in meters to nearest MRT station
* X4 number of convenience stores : count of nearby convenience stores
* X5 latitude : latitude in degrees
* X6 longitude : longitude in degrees
* Y house price of unit area : house price per unit area, in 10,000 New Taiwan Dollar per Ping
