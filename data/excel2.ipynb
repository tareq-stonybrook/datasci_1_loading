## Default of Credit Card Clients

Dataset on customer default payments in Taiwan, collected in 2005. Used to compare predictive accuracy of default probability across data mining methods. 30,000 instances, 23 explanatory variables plus a binary target.

## Columns

* ID : row index
* X1 (LIMIT_BAL) : amount of given credit (NT dollar), includes individual and family supplementary credit
* X2 (SEX) : gender, 1 = male, 2 = female
* X3 (EDUCATION) : education level, 1 = graduate school, 2 = university, 3 = high school, 4 = others
* X4 (MARRIAGE) : marital status, 1 = married, 2 = single, 3 = others
* X5 (AGE) : age in years
* X6 to X11 (PAY_0 to PAY_6) : repayment status from September to April 2005, scale is -1 = pay duly, 1 = delay one month, 2 = delay two months, up to 9 = delay nine months and above
* X12 to X17 (BILL_AMT1 to BILL_AMT6) : bill statement amount (NT dollar) from September to April 2005
* X18 to X23 (PAY_AMT1 to PAY_AMT6) : amount of previous payment (NT dollar) from September to April 2005
* Y (default payment next month) : target variable, 1 = default, 0 = no default
