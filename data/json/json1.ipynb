## Cars

Dataset of 406 cars from the 1970s to 1980s, used for exploring relationships between fuel efficiency and other vehicle attributes.

## Columns

* Name : car make and model
* Miles_per_Gallon : fuel efficiency in miles per gallon
* Cylinders : number of engine cylinders
* Displacement : engine displacement
* Horsepower : engine horsepower
* Weight_in_lbs : vehicle weight in pounds
* Acceleration : time to accelerate (seconds)
* Year : model year
* Origin : country of origin (USA, Europe, Japan)
