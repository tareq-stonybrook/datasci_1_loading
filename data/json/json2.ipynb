## Palmer Penguins Dataset

Dataset contains body measurements for 344 penguins across 3 species and 3 islands.

## Columns

1. **Species**: penguin species (Adelie, Chinstrap, Gentoo)
2. **Island**: island where penguin observed (Torgersen, Biscoe, Dream)
3. **Beak Length (mm)**: length of beak in millimeters
4. **Beak Depth (mm)**: depth of beak in millimeters
5. **Flipper Length (mm)**: flipper length in millimeters
6. **Body Mass (g)**: body mass in grams
7. **Sex**: MALE or FEMALE
