## Pima Indians Diabetes Dataset

Dataset contains medical diagnostic measurements for 768 female patients used to predict diabetes onset.

## Columns

1. **Pregnancies**: number of times pregnant
2. **Glucose**: plasma glucose concentration (2 hour oral glucose tolerance test)
3. **BloodPressure**: diastolic blood pressure (mm Hg)
4. **SkinThickness**: triceps skin fold thickness (mm)
5. **Insulin**: 2 hour serum insulin (mu U/ml)
6. **BMI**: body mass index
7. **DiabetesPedigreeFunction**: score reflecting diabetes likelihood based on family history
8. **Age**: age in years
9. **Outcome**: 1 indicates diabetes, 0 indicates no diabetes
